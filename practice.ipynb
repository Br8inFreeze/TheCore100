```python
from typing import List, Optional
import heapq
from collections import defaultdict, deque, Counter
import random

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def make_list(vals):
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def list_vals(head):
    res = []
    while head:
        res.append(head.val)
        head = head.next
    return res

def make_tree(vals):
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = deque([root])
    i = 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root

print("Setup complete")
```

## Binary Search (二分查找)

```python
# LC 704 - Binary Search
def search(nums: List[int], target: int) -> int:
    left, right = 0, len(nums)-1
    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            right = mid - 1
        else:
            left = mid + 1
    return -1

assert search([-1,0,3,5,9,12], 9) == 4
assert search([-1,0,3,5,9,12], 2) == -1
print("LC 704 passed")
```

```python
# LC 34 - Find First and Last Position of Element in Sorted Array
def searchRange(nums: List[int], target: int) -> List[int]:
    res_left = -1
    res_right = -1
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = (right + left) // 2
        if nums[mid] == target:
            res_left = mid
            right = mid - 1
        elif nums[mid] > target:
            right = mid - 1
        else:
            left = mid + 1
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = (right + left) // 2
        if nums[mid] == target:
            res_right = mid
            left = mid + 1
        elif nums[mid] > target:
            right = mid - 1
        else:
            left = mid + 1
    return [res_left, res_right]
                

assert searchRange([5,7,7,8,8,10], 8) == [3, 4]
assert searchRange([5,7,7,8,8,10], 6) == [-1, -1]
assert searchRange([], 0) == [-1, -1]
print("LC 34 passed")
```

```python
# LC 33 - Search in Rotated Sorted Array
def searchRotated(nums: List[int], target: int) -> int:
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = (right + left) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] > nums[right]:
            if nums[mid] > target:
                left = mid + 1
            else:
                right = mid - 1
        else:
            if nums[mid] > target:
                right = mid - 1
            else:
                left = mid + 1
    return -1

assert searchRotated([4,5,6,7,0,1,2], 0) == 4
assert searchRotated([4,5,6,7,0,1,2], 3) == -1
assert searchRotated([1], 0) == -1
print("LC 33 passed")
```

```python
# LC 162 - Find Peak Element
def findPeakElement(nums: List[int]) -> int:
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = (right + left) // 2
        if nums[mid] <= nums[mid + 1]:
            left = mid + 1
        else:
            right = mid - 1
    return left

assert findPeakElement([1,2,3,1]) == 2
assert findPeakElement([1,2,1,3,5,6,4]) in [1, 5]
print("LC 162 passed")
```
